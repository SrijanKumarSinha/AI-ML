Ex - 7.3 == Multiples of Ten

```python
number = input("Enter a number, and I'll tell you if it is multiple of 10 or not: ")
number = int(number)

if number % 10 == 0:
    print(f"The number {number} is multiple of 10")
else:
    print(f"The number {number} is not the multiple of 10")
```
